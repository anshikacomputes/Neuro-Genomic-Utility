```
# Project 1: Basic Genomic Motif Finder for Neuro-Research
# Author: Anshika Mishra

# We use a simulated sequence to demonstrate the tool's logic.
# In a real lab, this would be a FASTA file from a database like NCBI.
dna_sequence = "ATGCGATCGTAGCTAGCTAGCTAGCATGCTAGCTAGCTAGCTAGCTAGCTAGCT"

def analyze_sequence(seq):
    # Calculate GC Content: Higher GC content often implies higher thermal stability of DNA.
    gc_count = seq.count('G') + seq.count('C')
    gc_percent = (gc_count / len(seq)) * 100

    # Locate Start Codon (ATG): The point where translation usually begins.
    start_index = seq.find("ATG")

    # Search for a Regulatory Motif: Motifs are short patterns that suggest protein-binding sites.
    # I am looking for 'TAGCTA' as a hypothetical regulatory element.
    motif = "TAGCTA"
    motif_count = seq.count(motif)

    print(f"--- Molecular Analysis Results ---")
    print(f"Total Sequence Length: {len(seq)} bp")
    print(f"GC Content: {gc_percent:.2f}%")
    print(f"Start Codon found at position: {start_index}")
    print(f"Target Motif '{motif}' frequency: {motif_count}")

analyze_sequence(dna_sequence)
```
